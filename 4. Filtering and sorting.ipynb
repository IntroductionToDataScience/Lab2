**Hint**: A Boolean mask contains True and False values. In numerical operations, NumPy treats True as 1 and False as 0, so summing a Boolean mask counts how many elements satisfy the condition. Example
```Python
values = np.array([2, 7, 4, 9, 1])

values > 5
# [False, True, False, True, False]

np.sum(values > 5)
# 2
```
