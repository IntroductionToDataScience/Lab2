```python

```

Display all resulting arrays.

>Note (for the curious ones)

**Standardization transforms the values so that they are expressed in terms of their distance from the mean, measured in standard deviations.

- 0 means the value is equal to the mean,
- a positive value means it is above the mean,
- a negative value means it is below the mean,
- 1.5 means the value is 1.5 standard deviations above the mean.

After standardization, the data has a mean of approximately 0 and a standard deviation of approximately 1.**
